```python
#Broadcasting:
# the term broadcasting tells how numpy treats array with different shapes during arhmatics operation.
# the smaller array is 'broadcast' across the larger array so that they have compatible shapes.


#same shape:
a=np.arange(6).reshape(2,3)
b=np.arange(6,12).reshape(2,3)
print(a)
print(b)
print(a+b)#we can add array if they have same shape 

#diff shape;
a=np.arange(6).reshape(2,3)
b=np.arange(3).reshape(1,3)

print(a)
print(b)
print(a+b)

#BROADCASTING :
#1.Make two arrays have the same dimension ,i the no.of dimension of the two array are different,addnew dimewnsion with size 1 to the head of the array with the smaller dimension .
#2.make each dimension of the two array the same size,if the size of each dimension of the two array do not match,dimension with size 1,are streached to the size of the other array.
#if there is a dimension whose size is not 1 in either of the two array,it cannot be broadcasted,and an error is raised.

a=np.arange(12).reshape(4,3)
b=np.arange(3)
print(a)
print(b)
print(a+b)

c=np.arange(12).reshape(3,4)
d=np.arange(3)
print(c)
print(d)
#print(c+d) # THIS THROWS AN BECAUSE OF THE BROADCASTING RULES 

a=np.arange(3).reshape(1,3)
b=np.arange(3).reshape(3,1)
print(a)
print(b)
print(a+b)

d=np.arange(3).reshape(1,3)
e=np.arange(4).reshape(4,1)
print(d)
print(e)
print(d+e)

a=np.array([1])
b=np.arange(4).reshape(2,2)
print(a)
print(b)
print(a+b)

a=np.arange(12).reshape(3,4)
b=np.arange(12).reshape(4,3)
print(a)
print(b)
#print(a+b)# this will throw an error as there isnt'1' which broke broadcasting rules 

a=np.arange(16).reshape(4,4)
b=np.arange(4).reshape(2,2)
print(a)
print(b)
#print(a+b)#same there isnt 1 so it will not work 



```
